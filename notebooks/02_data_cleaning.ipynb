```python
import numpy as np
import pandas as pd
import os


pd.set_option("display.max_columns", 87)

train_path = "../data/raw/train.csv"

df = pd.read_csv(train_path)

df.head()
```

```python
missing_values = df.isnull().sum()

missing_values = missing_values[missing_values > 0].sort_values(ascending=False)

missing_values
```

```python
# Create a copy for the cleaned dataset
df_clean = df.copy()

df_clean.shape
```

```python
# Columns where NaN means the feature does not exist
none_columns = [
    "PoolQC",
    "MiscFeature",
    "Alley",
    "Fence",
    "FireplaceQu",
    "GarageType",
    "GarageFinish",
    "GarageQual",
    "GarageCond",
    "BsmtFinType1",
    "BsmtFinType2",
    "BsmtExposure",
    "BsmtCond",
    "BsmtQual",
    "MasVnrType"
]

df_clean[none_columns] = df_clean[none_columns].fillna("None")

# Check the result
df_clean[none_columns].isnull().sum()
```

```python
# No masonry veneer means zero masonry veneer area
df_clean["MasVnrArea"] = df_clean["MasVnrArea"].fillna(0)

# Verify
df_clean["MasVnrArea"].isnull().sum()
```

```python
# Fill missing LotFrontage using the median of each neighborhood
df_clean["LotFrontage"] = (
    df_clean.groupby("Neighborhood")["LotFrontage"]
    .transform(lambda x: x.fillna(x.median()))
)

# Verify
df_clean["LotFrontage"].isnull().sum()
```

```python
# Find the row with missing Electrical
df[df["Electrical"].isnull()]
```

```python
# Compare electrical systems among similar houses
timber_comparison = df[
    (df["Neighborhood"] == "Timber") &
    (df["OverallQual"].between(5, 6))
][["Id", "Neighborhood", "OverallQual", "YearBuilt", "Electrical", "SalePrice"]]

timber_comparison.sort_values("OverallQual")
```

```python
# Impute the single missing Electrical value
df_clean["Electrical"] = df_clean["Electrical"].fillna("SBrkr")

# Verify
df_clean["Electrical"].isnull().sum()
```

```python
# Check for any remaining missing values
remaining_missing = df_clean.isnull().sum()

remaining_missing = remaining_missing[remaining_missing > 0].sort_values(ascending=False)

remaining_missing
```

```python
# No garage means there is no garage construction year
df_clean["GarageYrBlt"] = df_clean["GarageYrBlt"].fillna(0)

# Verify
df_clean["GarageYrBlt"].isnull().sum()
```

```python
# Final missing-value verification
remaining_missing = df_clean.isnull().sum()

remaining_missing = remaining_missing[remaining_missing > 0].sort_values(ascending=False)

remaining_missing
```

```python
# Check data types after cleaning
df_clean.dtypes
```

```python
# Identify numeric and categorical columns
numeric_columns = df_clean.select_dtypes(include=["int64", "float64"]).columns
categorical_columns = df_clean.select_dtypes(include=["object"]).columns

print("Numeric columns:", len(numeric_columns))
print("Categorical columns:", len(categorical_columns))
```

```python
# Check for duplicate rows
duplicate_count = df_clean.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
# Summary statistics for numeric variables
df_clean[numeric_columns].describe().T
```

```python
# Inspect houses with the largest above-ground living areas
df_clean[
    ["Id", "Neighborhood", "OverallQual", "GrLivArea", "SalePrice"]
].sort_values("GrLivArea", ascending=False).head(10)
```

```python
# Inspect the two most unusual large properties
df_clean[
    (df_clean["Id"].isin([1299, 524]))
][[
    "Id",
    "Neighborhood",
    "OverallQual",
    "OverallCond",
    "YearBuilt",
    "YearRemodAdd",
    "GrLivArea",
    "TotalBsmtSF",
    "1stFlrSF",
    "2ndFlrSF",
    "GarageCars",
    "GarageArea",
    "FullBath",
    "BedroomAbvGr",
    "SalePrice"
]]
```

```python
# Check logical consistency between key house features

consistency_checks = {
    "GrLivArea": (
        df_clean["GrLivArea"]
        == df_clean["1stFlrSF"] + df_clean["2ndFlrSF"] + df_clean["LowQualFinSF"]
    ).sum(),

    "TotalBsmtSF": (
        df_clean["TotalBsmtSF"]
        == df_clean["BsmtFinSF1"] + df_clean["BsmtFinSF2"] + df_clean["BsmtUnfSF"]
    ).sum(),

    "GarageArea_zero_with_no_cars": (
        (df_clean["GarageCars"] == 0) & (df_clean["GarageArea"] != 0)
    ).sum(),

    "GarageCars_zero_with_no_garage": (
        (df_clean["GarageCars"] == 0) & (df_clean["GarageType"] != "None")
    ).sum(),

    "PoolArea_zero_with_pool_quality": (
        (df_clean["PoolArea"] == 0) & (df_clean["PoolQC"] != "None")
    ).sum()
}

consistency_checks
```

```python
# Check logical consistency of year variables

year_checks = {
    "YearRemodAdd_before_YearBuilt": (
        df_clean["YearRemodAdd"] < df_clean["YearBuilt"]
    ).sum(),

    "YearBuilt_after_YrSold": (
        df_clean["YearBuilt"] > df_clean["YrSold"]
    ).sum(),

    "GarageYrBlt_after_YrSold": (
        (df_clean["GarageYrBlt"] > df_clean["YrSold"]) &
        (df_clean["GarageYrBlt"] != 0)
    ).sum(),

    "GarageYrBlt_before_YearBuilt": (
        (df_clean["GarageYrBlt"] < df_clean["YearBuilt"]) &
        (df_clean["GarageYrBlt"] != 0)
    ).sum()
}

year_checks
```

```python
# Inspect properties where the garage year predates the house construction year

garage_year_issues = df_clean[
    (df_clean["GarageYrBlt"] < df_clean["YearBuilt"]) &
    (df_clean["GarageYrBlt"] != 0)
][[
    "Id",
    "Neighborhood",
    "YearBuilt",
    "YearRemodAdd",
    "GarageYrBlt",
    "GarageType",
    "GarageCars",
    "GarageArea",
    "OverallQual",
    "SalePrice"
]].sort_values("GarageYrBlt")

garage_year_issues
```

```python
# Final validation of the cleaned dataset

print("Dataset shape:", df_clean.shape)
print("Total missing values:", df_clean.isnull().sum().sum())
print("Duplicate rows:", df_clean.duplicated().sum())
print("Numeric columns:", len(df_clean.select_dtypes(include=["int64", "float64"]).columns))
print("Categorical columns:", len(df_clean.select_dtypes(include=["object"]).columns))
print("Target missing values:", df_clean["SalePrice"].isnull().sum())
```

```python


# Create processed data directory
os.makedirs("../data/processed", exist_ok=True)

# Save cleaned dataset
cleaned_path = "../data/processed/cleaned_train.csv"
df_clean.to_csv(cleaned_path, index=False)

print(f"Cleaned dataset saved to: {cleaned_path}")
```

```python
# Reload the saved cleaned dataset
df_saved = pd.read_csv(cleaned_path)

print("Saved dataset shape:", df_saved.shape)
print("Saved dataset missing values:", df_saved.isnull().sum().sum())
print("Saved dataset duplicates:", df_saved.duplicated().sum())

# Verify that the saved data matches the cleaned dataframe
print("Data matches cleaned dataframe:", df_saved.equals(df_clean))
```
