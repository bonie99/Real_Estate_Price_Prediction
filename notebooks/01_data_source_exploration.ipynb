```python
# Identify categorical columns
categorical_columns = df.select_dtypes(include=["object"]).columns

print("Number of categorical columns:", len(categorical_columns))
print("\nCategorical columns:")
print(categorical_columns.tolist())
```

```python
# Summary of sale prices by selected categorical variables

categorical_features = [
    "Neighborhood",
    "MSZoning",
    "HouseStyle",
    "Foundation",
    "HeatingQC",
    "KitchenQual",
    "GarageFinish",
    "BsmtQual",
    "ExterQual",
    "SaleCondition"
]

for col in categorical_features:
    print(f"\n{'='*60}")
    print(f"{col}")
    print(f"{'='*60}")
    
    summary = (
        df.groupby(col)["SalePrice"]
        .agg(["count", "mean", "median"])
        .sort_values("median", ascending=False)
    )
    
    print(summary)
```

```python
# Compare sale prices across important categorical variables

categorical_features = [
    "Neighborhood",
    "MSZoning",
    "HouseStyle",
    "Foundation",
    "HeatingQC",
    "KitchenQual",
    "GarageFinish",
    "BsmtQual",
    "ExterQual",
    "SaleCondition"
]

for col in categorical_features:
    print(f"\n{'='*60}")
    print(f"{col}")
    print(f"{'='*60}")
    
    summary = (
        df.groupby(col)["SalePrice"]
        .agg(["count", "mean", "median","min","max"])
        .sort_values("median", ascending=False)
    )
    
    print(summary)
```
